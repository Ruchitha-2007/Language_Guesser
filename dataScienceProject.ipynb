```python
%%writefile extract_features.py
import os
import cv2
import numpy as np
import mediapipe as mp
from tqdm import tqdm
import shutil

# ==========================================
# CONFIGURATION
# ==========================================
# Local fast paths
DATASET_RAW_DIR = "/content/dataset_raw"
DATASET_OUT_DIR = "/content/dataset_landmarks"
SEQUENCE_LENGTH = 30

# Google Drive Destination for your friends
DRIVE_FOLDER = "/content/drive/MyDrive/ISL_Project"
DRIVE_ZIP_PATH = os.path.join(DRIVE_FOLDER, "dataset_landmarks.zip")

UPPER_BODY_LANDMARKS = list(range(25))
mp_holistic = mp.solutions.holistic

def extract_raw_frame_landmarks(results):
    pose_raw = None
    if results.pose_landmarks:
        pose_raw = np.array([[lm.x, lm.y, lm.z] for lm in results.pose_landmarks.landmark], dtype=np.float32)

    lh_raw = None
    if results.left_hand_landmarks:
        lh_raw = np.array([[lm.x, lm.y, lm.z] for lm in results.left_hand_landmarks.landmark], dtype=np.float32)

    rh_raw = None
    if results.right_hand_landmarks:
        rh_raw = np.array([[lm.x, lm.y, lm.z] for lm in results.right_hand_landmarks.landmark], dtype=np.float32)

    return pose_raw, lh_raw, rh_raw

def fill_missing_hands_temporally(hand_sequence):
    total_frames = len(hand_sequence)
    valid_indices = [i for i, h in enumerate(hand_sequence) if h is not None]

    if not valid_indices:
        return np.zeros((total_frames, 21, 3), dtype=np.float32)

    filled_array = np.zeros((total_frames, 21, 3), dtype=np.float32)
    for i, h in enumerate(hand_sequence):
        if h is not None:
            filled_array[i] = h

    first_valid = valid_indices[0]
    for i in range(0, first_valid):
        filled_array[i] = filled_array[first_valid]

    for i in range(first_valid + 1, total_frames):
        if hand_sequence[i] is None:
            filled_array[i] = filled_array[i - 1]

    return filled_array

def process_video_robust(video_path, holistic):
    cap = cv2.VideoCapture(video_path)
    pose_history = []
    lh_history = []
    rh_history = []

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        image_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        results = holistic.process(image_rgb)

        pose_raw, lh_raw, rh_raw = extract_raw_frame_landmarks(results)
        pose_history.append(pose_raw)
        lh_history.append(lh_raw)
        rh_history.append(rh_raw)

    cap.release()

    total_raw_frames = len(pose_history)
    if total_raw_frames < 5:
        return None

    valid_pose_idx = [i for i, p in enumerate(pose_history) if p is not None]
    if not valid_pose_idx:
        return None

    filled_pose = np.zeros((total_raw_frames, 33, 3), dtype=np.float32)
    for i, p in enumerate(pose_history):
        if p is not None:
            filled_pose[i] = p
        elif i > 0:
            filled_pose[i] = filled_pose[i-1]
        else:
            filled_pose[i] = pose_history[valid_pose_idx[0]]

    filled_lh = fill_missing_hands_temporally(lh_history)
    filled_rh = fill_missing_hands_temporally(rh_history)

    normalized_sequence = []
    for i in range(total_raw_frames):
        pose_f = filled_pose[i]
        lh_f = filled_lh[i]
        rh_f = filled_rh[i]

        nose = pose_f[0]
        shoulder_dist = np.linalg.norm(pose_f[11] - pose_f[12]) + 1e-6

        # Pose
        pose_norm = (pose_f - nose) / shoulder_dist
        pose_upper = pose_norm[UPPER_BODY_LANDMARKS].flatten()

        # Hands
        if np.any(lh_f != 0):
            lh_norm = ((lh_f - nose) / shoulder_dist).flatten()
        else:
            lh_norm = np.zeros(21 * 3, dtype=np.float32)

        if np.any(rh_f != 0):
            rh_norm = ((rh_f - nose) / shoulder_dist).flatten()
        else:
            rh_norm = np.zeros(21 * 3, dtype=np.float32)

        combined_frame = np.concatenate([pose_upper, lh_norm, rh_norm])
        normalized_sequence.append(combined_frame)

    indices = np.linspace(0, total_raw_frames - 1, SEQUENCE_LENGTH, dtype=int)
    final_sequence = np.array([normalized_sequence[idx] for idx in indices], dtype=np.float32)

    return final_sequence

def build_dataset():
    os.makedirs(DATASET_OUT_DIR, exist_ok=True)
    os.makedirs(DRIVE_FOLDER, exist_ok=True)

    classes = sorted([d for d in os.listdir(DATASET_RAW_DIR) if os.path.isdir(os.path.join(DATASET_RAW_DIR, d))])
    print(f"Total classes to process: {len(classes)}")

    with mp_holistic.Holistic(min_detection_confidence=0.5, min_tracking_confidence=0.5) as holistic:
        for class_name in tqdm(classes, desc="Overall Classes"):
            class_raw_path = os.path.join(DATASET_RAW_DIR, class_name)
            class_out_path = os.path.join(DATASET_OUT_DIR, class_name)
            os.makedirs(class_out_path, exist_ok=True)

            video_files = [f for f in os.listdir(class_raw_path) if f.lower().endswith(('.mp4', '.avi', '.mov'))]

            for idx, video_file in enumerate(video_files):
                out_file = os.path.join(class_out_path, f"sample_{idx:03d}.npy")
                if os.path.exists(out_file):
                    continue

                video_path = os.path.join(class_raw_path, video_file)
                data = process_video_robust(video_path, holistic)
                if data is not None:
                    np.save(out_file, data)

    # Automatically zip and save to Google Drive
    print(f"\nCompressing and saving to Google Drive: {DRIVE_ZIP_PATH} ...")
    shutil.make_archive(base_name=os.path.splitext(DRIVE_ZIP_PATH)[0], format='zip', root_dir=DATASET_OUT_DIR)
    print("Done! The preprocessed dataset is now permanently saved in your Google Drive.")

if __name__ == "__main__":
    build_dataset()
```

```python
import requests
import os
from tqdm import tqdm

# Query Zenodo record 4010759
record_id = "4010759"
url = f"https://zenodo.org/api/records/{record_id}"
headers = {"User-Agent": "Mozilla/5.0"}

response = requests.get(url, headers=headers)
data = response.json()

files = data.get("files", [])
print(f"Found {len(files)} files on Zenodo:\n")

for f in files:
    filename = f["key"]
    download_url = f["links"]["self"]
    size_mb = f["size"] / (1024 * 1024)
    print(f"File: {filename} ({size_mb:.2f} MB)")

    # Download file with progress bar
    if not os.path.exists(filename):
        print(f"Downloading {filename}...")
        with requests.get(download_url, headers=headers, stream=True) as r:
            r.raise_for_status()
            total_size = int(r.headers.get("content-length", 0))
            with open(filename, "wb") as out_f, tqdm(
                total=total_size, unit="B", unit_scale=True, desc=filename
            ) as bar:
                for chunk in r.iter_content(chunk_size=8192 * 16):
                    out_f.write(chunk)
                    bar.update(len(chunk))
    else:
        print(f"{filename} already exists. Skipping.")
```

```python
%%writefile extract_features.py
import os
import cv2
import numpy as np
import mediapipe as mp
from tqdm import tqdm

# ==========================================
# CONFIGURATION
# ==========================================
DATASET_RAW_DIR = "dataset_raw"
DATASET_OUT_DIR = "dataset_landmarks"
SEQUENCE_LENGTH = 30

UPPER_BODY_LANDMARKS = list(range(25))
mp_holistic = mp.solutions.holistic

def extract_raw_frame_landmarks(results):
    pose_raw = None
    if results.pose_landmarks:
        pose_raw = np.array([[lm.x, lm.y, lm.z] for lm in results.pose_landmarks.landmark], dtype=np.float32)

    lh_raw = None
    if results.left_hand_landmarks:
        lh_raw = np.array([[lm.x, lm.y, lm.z] for lm in results.left_hand_landmarks.landmark], dtype=np.float32)

    rh_raw = None
    if results.right_hand_landmarks:
        rh_raw = np.array([[lm.x, lm.y, lm.z] for lm in results.right_hand_landmarks.landmark], dtype=np.float32)

    return pose_raw, lh_raw, rh_raw

def fill_missing_hands_temporally(hand_sequence):
    total_frames = len(hand_sequence)
    valid_indices = [i for i, h in enumerate(hand_sequence) if h is not None]

    if not valid_indices:
        return np.zeros((total_frames, 21, 3), dtype=np.float32)

    filled_array = np.zeros((total_frames, 21, 3), dtype=np.float32)
    for i, h in enumerate(hand_sequence):
        if h is not None:
            filled_array[i] = h

    first_valid = valid_indices[0]
    for i in range(0, first_valid):
        filled_array[i] = filled_array[first_valid]

    for i in range(first_valid + 1, total_frames):
        if hand_sequence[i] is None:
            filled_array[i] = filled_array[i - 1]

    return filled_array

def process_video_robust(video_path, holistic):
    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        cap.release()
        return None

    pose_history = []
    lh_history = []
    rh_history = []

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        image_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
        results = holistic.process(image_rgb)

        pose_raw, lh_raw, rh_raw = extract_raw_frame_landmarks(results)
        pose_history.append(pose_raw)
        lh_history.append(lh_raw)
        rh_history.append(rh_raw)

    cap.release()

    total_raw_frames = len(pose_history)
    if total_raw_frames < 5:
        return None

    valid_pose_idx = [i for i, p in enumerate(pose_history) if p is not None]
    if not valid_pose_idx:
        return None

    filled_pose = np.zeros((total_raw_frames, 33, 3), dtype=np.float32)
    for i, p in enumerate(pose_history):
        if p is not None:
            filled_pose[i] = p
        elif i > 0:
            filled_pose[i] = filled_pose[i-1]
        else:
            filled_pose[i] = pose_history[valid_pose_idx[0]]

    filled_lh = fill_missing_hands_temporally(lh_history)
    filled_rh = fill_missing_hands_temporally(rh_history)

    normalized_sequence = []
    for i in range(total_raw_frames):
        pose_f = filled_pose[i]
        lh_f = filled_lh[i]
        rh_f = filled_rh[i]

        nose = pose_f[0]
        shoulder_dist = np.linalg.norm(pose_f[11] - pose_f[12]) + 1e-6

        pose_norm = (pose_f - nose) / shoulder_dist
        pose_upper = pose_norm[UPPER_BODY_LANDMARKS].flatten()

        if np.any(lh_f != 0):
            lh_norm = ((lh_f - nose) / shoulder_dist).flatten()
        else:
            lh_norm = np.zeros(21 * 3, dtype=np.float32)

        if np.any(rh_f != 0):
            rh_norm = ((rh_f - nose) / shoulder_dist).flatten()
        else:
            rh_norm = np.zeros(21 * 3, dtype=np.float32)

        combined_frame = np.concatenate([pose_upper, lh_norm, rh_norm])
        normalized_sequence.append(combined_frame)

    indices = np.linspace(0, total_raw_frames - 1, SEQUENCE_LENGTH, dtype=int)
    final_sequence = np.array([normalized_sequence[idx] for idx in indices], dtype=np.float32)

    return final_sequence

def build_dataset():
    os.makedirs(DATASET_OUT_DIR, exist_ok=True)
    classes = sorted([d for d in os.listdir(DATASET_RAW_DIR) if os.path.isdir(os.path.join(DATASET_RAW_DIR, d))])
    print(f"Total classes to process: {len(classes)}")

    with mp_holistic.Holistic(min_detection_confidence=0.5, min_tracking_confidence=0.5) as holistic:
        for class_name in tqdm(classes, desc="Overall Classes"):
            class_raw_path = os.path.join(DATASET_RAW_DIR, class_name)
            class_out_path = os.path.join(DATASET_OUT_DIR, class_name)
            os.makedirs(class_out_path, exist_ok=True)

            # Case-insensitive check for .mp4, .avi, .mov (including .MOV)
            video_files = [f for f in os.listdir(class_raw_path) if f.lower().endswith(('.mp4', '.avi', '.mov'))]

            for idx, video_file in enumerate(video_files):
                out_file = os.path.join(class_out_path, f"sample_{idx:03d}.npy")
                if os.path.exists(out_file):
                    continue

                video_path = os.path.join(class_raw_path, video_file)
                data = process_video_robust(video_path, holistic)
                if data is not None:
                    np.save(out_file, data)

if __name__ == "__main__":
    build_dataset()
```
